```python
import torch
import torch.nn.functional as F
from matplotlib import pyplot as plt
```

```python
logits = torch.randn(20, 32)
# Sample soft categorical using reparametrization trick:
gsm = F.gumbel_softmax(logits, tau=0.001, hard=False, dim=-1)
```

```python
fig, ax = plt.subplots(1,2, figsize=(10, 5))
ax[0].imshow(logits.detach().numpy(), aspect='auto')
ax[1].imshow(gsm.detach().numpy(), aspect='auto')
plt.show()
```

```python
i = 1
plt.plot(logits[i].detach().numpy())
plt.plot(gsm[i].detach().numpy())
plt.legend(['logits', 'gsm'])
plt.show()
```

```python
# Sample hard categorical using "Straight-through" trick:
F.gumbel_softmax(logits, tau=1, hard=True)
```
