```python
### ALL lr

# ===== Your settings =====
actor_lrs  = [0.0003, 0.001, 0.003]
critic_lrs = [0.0015, 0.005, 0.015]
configurations = ['simple', 'eisep_actor', 'eisep_critic', 'eisep_all']  # top->bottom order

# out_tex  = Path("./rl/result/lunarlander/simple_rnn_ei_eisep/a2c_lunarlander_table.tex")
out_tex = Path(f"./rl/result/{task_name}/simple_{simple_type}_ei_{ei_type}/a2c_{task_name}_lr_sweep_table.tex")

# ===== Load & aggregate =====
reward_ars = np.load(npy_path)
reward_ar_flatten = reward_ars.reshape(
    len(actor_lrs), len(critic_lrs), len(configurations), -1
)
mean_r = reward_ar_flatten.mean(axis=-1)
ste_r  = reward_ar_flatten.std(axis=-1, ddof=0) / np.sqrt(reward_ar_flatten.shape[-1])

config_names   = ['Simple', 'EISep(A)', 'EISep(C)', 'EISep(A+C)']  # for caption text

# Input / output
out_tex  = Path(f"./rl/result/{task_name}/simple_{simple_type}_ei_{ei_type}/a2c_{task_name}_lr_sweep_table.tex")

# Seeds description (for caption)
N_TRAIN_SEEDS = 5
N_TEST_SEEDS  = 20


A, C, K = mean_r.shape
assert K == len(configurations), f"K mismatch: mean_r has K={K}, but configurations has {len(configurations)}"

# ==================================
# ===== Formatting helpers ==========
# ==================================
def fmt_num(x: float) -> str:
    return f"{x:.2f}"

def fmt_lr(x: float) -> str:
    """Return scientific notation in LaTeX form, e.g., 3e-4 -> 3\\times10^{-4}."""
    x = float(x)
    if x == 0:
        return "0"
    exp = int(np.floor(np.log10(abs(x))))
    base = x / (10 ** exp)
    # Use g to avoid trailing zeros, keep it compact
    return f"{base:g}\\times10^{{{exp}}}"

def bold_math(math_text: str) -> str:
    """
    Input like '$12.34 \\pm 0.56$' and return bold math.
    We keep it in math mode and apply \\mathbf to the whole content.
    """
    t = math_text.strip()
    if t.startswith("$") and t.endswith("$"):
        t = t[1:-1]
    return f"$\\mathbf{{{t}}}$"

# ==========================================================
# ===== Find best (actor_lr, critic_lr) per configuration ====
# ==========================================================
# For each model k, find the lr pair (i,j) that maximizes mean reward.
# If ties happen, we bold all cells achieving the same maximum (within tolerance).
best_mask = np.zeros((A, C, K), dtype=bool)
for k in range(K):
    mx = np.max(mean_r[:, :, k])
    # tolerance for floating comparisons
    best_mask[:, :, k] = np.isclose(mean_r[:, :, k], mx, rtol=1e-12, atol=1e-12)

# =========================================
# ===== Build one table cell (4 lines) =====
# =========================================
def cell_block(i_actor: int, j_critic: int) -> str:
    """
    Create a 4-line makecell. For each config k:
      show mean±SE, and bold it if (i_actor, j_critic) is that config's best lr pair.
    """
    lines = []
    for k in range(K):
        m = fmt_num(mean_r[i_actor, j_critic, k])
        s = fmt_num(ste_r[i_actor, j_critic, k])
        txt = f"${m} \\pm {s}$"
        if best_mask[i_actor, j_critic, k]:
            txt = bold_math(txt)
        lines.append(txt)
    return "\\makecell[tc]{" + "\\\\[1pt]".join(lines) + "}"

# =========================
# ===== Build LaTeX =======
# =========================
lines = []
lines.append("\\begin{table}[t]")
lines.append("  \\centering")
lines.append("  \\caption{Learning-rate sweep on LunarLander. Each cell reports test return (mean $\\pm$ SE) for four model variants (top to bottom): "
             + ", ".join(config_names)
             + f". Boldface marks the best learning-rate pair for each model (searched over the whole grid). "
             + f"Each configuration uses $n={N_TRAIN_SEEDS}$ training seeds and each trained model is evaluated with $n={N_TEST_SEEDS}$ test seeds."
             + "}")
lines.append("  \\label{tab:lunarlander_lr_sweep}")
lines.append("  \\setlength{\\tabcolsep}{3pt}")
lines.append("  \\renewcommand{\\arraystretch}{0.9}")
lines.append("  \\scriptsize")
lines.append("  \\begin{tabular}{l" + "c"*len(actor_lrs) + "}")
lines.append("    \\toprule")

actor_hdr = " & ".join([f"$\\mathbf{{{fmt_lr(a)}}}$" for a in actor_lrs])
lines.append(f"    \\textbf{{lr(C) $\\backslash$ lr(A)}} & {actor_hdr} \\\\")
lines.append("    \\midrule")

for j, c_lr in enumerate(critic_lrs):
    row_cells = [f"$\\mathbf{{{fmt_lr(c_lr)}}}$"]
    for i in range(len(actor_lrs)):
        row_cells.append(cell_block(i, j))
    lines.append("    " + " & ".join(row_cells) + " \\\\")
    if j < len(critic_lrs) - 1:
        lines.append("    \\midrule")

lines.append("    \\bottomrule")
lines.append("  \\end{tabular}")
lines.append("")
lines.append("  \\vspace{3pt}")
lines.append("  \\textit{Order in each cell (top to bottom):} "
             + ", ".join(config_names) + ".")
lines.append("\\end{table}")

out_tex.parent.mkdir(parents=True, exist_ok=True)
out_tex.write_text("\n".join(lines), encoding="utf-8")
print(out_tex.resolve())
print('./'+str(out_tex))
```

```python
import pickle as pkl
import gymnasium as gym
from net.a2c import A2C
def score(weight, module):
    sparsity = 1 - np.mean(np.abs(weight)>0.1*np.std(np.abs(weight)))
    exc_ratio = -1
    if hasattr(module, 'get_ei_entropy_loss'):
        eimask_logit = module.eimask_logit.detach().cpu().numpy()
        exc_ratio = np.mean(eimask_logit[:,0]>eimask_logit[:,1])
    return sparsity, exc_ratio
def sparsity_analysis(l1_coefs, critic_lr, actor_lr, simple_type, ei_type, task_name):
    # critic_lr = 0.005
    # actor_lr = 0.001
    hidden_size = 64
    env_name = "LunarLander-v3"
    # colors = ['#FA7F72', '#A84499', '#183319', '#20B2AA']
    # config_names = ['Simple', 'EISep(A)', 'EISep(C)', 'EISep(A+C)']
    # configuration = 'eisep_all'  # options: 'simple', 'eisep_actor', 'eisep_critic', 'eisep_all'
    # Note: the actor has a slower learning rate so that the value targets become
    # more stationary and are theirfore easier to estimate for the critic
    # environment hyperparams
    n_seed = 5
    n_envs = 1
    hidden_size = 64
    configuration = 'simple'
    env = gym.make(env_name, max_episode_steps=500)
    obs_shape = env.observation_space.shape[0]
    action_shape = env.action_space.n
    sparsity_all = np.zeros((len(l1_coefs), n_seed, 2))
    for i, l1_coef in enumerate(l1_coefs):
        info_path = f'./rl/result/{task_name}/L1_{l1_coef}_simple_{simple_type}_ei_{ei_type}/hidden_{hidden_size}/critic_{critic_lr}_actor_{actor_lr}/{configuration}.pkl'
        with open(info_path, 'rb') as f:
            info = pkl.load(f)
        net_state_dicts = info['net_state_dict']
        # environment setup
        device = 'cuda'
        # sparsity = np.zeros((n_seed, 2))
        for seed, state_dict in enumerate(net_state_dicts):
            # agent = A2C_singleRNN(obs_shape, action_shape, device, critic_lr, actor_lr, n_envs, hidden_size, configuration)
            agent = A2C(obs_shape, action_shape, device, critic_lr, actor_lr, n_envs, hidden_size, configuration,
                        simple_type, ei_type)
            agent.load_state_dict(net_state_dicts[seed])
            agent.eval()
            agent.to(device)
            # plot_recurrent_weight(agent)
            # W = agent.critic_actor.get_actual_weight().detach().cpu().numpy()
            W_critic = agent.critic.get_actual_weight().detach().cpu().numpy()
            W_actor = agent.actor.get_actual_weight().detach().cpu().numpy()
            sparsity_critic, exc_ratio_critic = score(W_critic, agent.critic)
            sparsity_actor, exc_ratio_actor = score(W_actor, agent.actor)
            sparsity_all[i, seed, 0] = sparsity_critic
            sparsity_all[i, seed, 1] = sparsity_actor
    return sparsity_all

actor_lrs = [0.0003, 0.001, 0.003]
critic_lrs = [0.0015, 0.005, 0.015]
l1_coefs = [0.001, 0.01, 0.1, 1, 10, 100, 1000]
n_showcase_episodes = 20
n_seed = 5
sparsity_ars = np.zeros((len(actor_lrs), len(critic_lrs), len(l1_coefs), n_seed, 2))
simple_type = 'rnn'
ei_type = 'eisep'
task_name = 'lunarlander'
for a, actor_lr in enumerate(actor_lrs):
    for c, critic_lr in enumerate(critic_lrs):
        print(f"actor_lr: {actor_lr}, critic_lr: {critic_lr}")
        sparsity_ars[a,c] = sparsity_analysis(l1_coefs, critic_lr, actor_lr, simple_type, ei_type, task_name)


from rl.rl_util import savefig
sparsity_flatten = sparsity_ars.transpose(2,0,1,3,4).reshape(len(l1_coefs), -1, 2)

# config_names = ['Simple', 'EISep(A)', 'EISep(C)', 'EISep(A+C)']
fig, ax = plt.subplots(1,1,figsize=(6,3), constrained_layout=True)
fontsize = 15
# x = np.arange(len(l1_coefs))
width = 0.35
critic_means = np.mean(sparsity_flatten[:,:,0], axis=1)
critic_stes = np.std(sparsity_flatten[:,:,0], axis=1)/np.sqrt(sparsity_flatten.shape[1])
actor_means = np.mean(sparsity_flatten[:,:,1], axis=1)
actor_stes = np.std(sparsity_flatten[:,:,1], axis=1)/np.sqrt(sparsity_flatten.shape[1])
# ax.bar(x - width/2, critic_means, width, yerr=critic_stes, label='Critic', capsize=4, color='#2965FF')
# ax.bar(x + width/2, actor_means, width, yerr=actor_stes, label='Actor', capsize=4, color='#C73D3D')
eiboth_sparsity_dict = pkl.load(open('./rl/output/lunarlander/simple_rnn_ei_eisep/sparsity_averaged_across_lr.pkl', 'rb'))
critic_mean_eiboth = eiboth_sparsity_dict['critic_means'][-1]
actor_mean_eiboth = eiboth_sparsity_dict['actor_means'][-1]
ax.plot(l1_coefs, critic_means, label='Critic', color='#2965FF',
        marker='o', markersize=10, markeredgecolor='white', markeredgewidth=1)
ax.fill_between(l1_coefs, critic_means - critic_stes, critic_means + critic_stes, alpha=0.2, color='#2965FF')
ax.plot(l1_coefs, actor_means, label='Actor', color='#C73D3D',
        marker='o', markersize=10, markeredgecolor='white', markeredgewidth=1)
ax.fill_between(l1_coefs, actor_means - actor_stes, actor_means + actor_stes, alpha=0.2, color='#C73D3D')
ax.axhline(critic_mean_eiboth, color='#2965FF', linestyle='dashed', linewidth=2)
ax.axhline(actor_mean_eiboth,  color='#C73D3D', linestyle='dashed', linewidth=2)
# ax.set_xticks(l1_coefs)
# ax.set_xticklabels(l1_coefs, fontsize=fontsize, rotation=0)
ax.set_xscale('log')
ax.set_ylim(0,1)
ax.set_ylabel('Sparsity', fontsize=fontsize+2)
ax.set_xlabel('L1 coefficient', fontsize=fontsize+2)
ax.set_title('Reinforcement learning sparsity', fontsize=fontsize+2)
ax.tick_params(labelsize=fontsize-1)
# ax.legend(fontsize=fontsize, frameon=False, loc='lower left')
# plt.tight_layout()
# plt.show()
savefig(f'./rl/output/lunarlander/multil1_sparsity_averaged_across_lr.png', dpi=400)
```
