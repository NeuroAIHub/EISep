```python
import warnings

import matplotlib.pyplot as plt
import numpy as np

from IPython.display import clear_output
warnings.filterwarnings("ignore")
clear_output()
```

```python
import json
import os
root = './ARC-AGI-2/data/training'
filename = root + '/' + os.listdir(root)[3]
print(filename)
file = open(filename)
json_file = json.load(file)
print(json_file)
```

```python
from matplotlib.colors import ListedColormap, Normalize
cmap = ListedColormap([
    '#000', '#0074D9', '#FF4136', '#2ECC40', '#FFDC00',
    '#AAAAAA', '#F012BE', '#FF851B', '#7FDBFF', '#870C25'
])
norm = Normalize(vmin=0, vmax=9)
print('train')
for i, example in enumerate(json_file['train']):
    train_input = np.array(example['input'])
    train_output = np.array(example['output'])
    # print(np.unique(train_input))
    print(train_input.shape)
    # print(np.unique(train_output))
    print(train_output.shape)
    fig, ax = plt.subplots(1,2,figsize=(6,3))
    ax[0].imshow(train_input, cmap='tab10', norm=norm)
    ax[1].imshow(train_output, cmap='tab10', norm=norm)
    plt.show()
    plt.close()
print('test')
for i, example in enumerate(json_file['test']):
    test_input = np.array(example['input'])
    test_output = np.array(example['output'])
    print(test_input.shape)
    print(test_output.shape)
    fig, ax = plt.subplots(1,2,figsize=(6,3))
    ax[0].imshow(test_input, cmap='tab10', vmin=0, vmax=9)
    ax[1].imshow(test_output, cmap='tab10', vmin=0, vmax=9)
    plt.show()
    plt.close()
```
